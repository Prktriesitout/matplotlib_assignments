A plot is the basic visualizatiom element that helps to visualize the data

Figure:- figure is the top - level container that acts as a window or page on which everything is drawn

Axes:- The Axes are the area on which data is plotted. it can have labels or ticks associated with it

Title:- the title is the name of the plot

Legend : A legend is a set fof key-value pairs that contains the list of plots and their respective labels

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([1,2,3,4,5])
y=x**2
z=(y/2)+40
```

```python

fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax2=fig.add_axes([0.2,0.5,0.4,0.3])
ax.plot(x,y,color='g', label='line1')
ax2.plot(x,z,color='r', label='line2')
ax.legend()
ax2.legend()
ax.set_title('Title of the plot',fontsize=15)
ax2.set_title('Title of the plot',fontsize=15)
ax.set_xlabel('x-axis',fontsize=15)

ax.set_ylabel('y-axis',fontsize=15)
plt.show()

```

```python
x=np.arange(1,16)
y=x**2
z=(y/2)+40
fig,(ax1,ax2)=plt.subplots(1,2)
ax1.plot(x,y,color="g",label='line1')
ax1.set_xlabel('Plot 1-X axis')
ax1.set_ylabel('Plot 1-Y axis')

ax2.plot(x,z,color="r",label='line2')
ax2.set_xlabel('Plot 2-X axis')
ax2.set_ylabel('Plot 2-Y axis')
plt.show()
fig.tight_layout()
```

```python
import pandas as pd
```

```python
df=pd.read_csv('auto_mpg.csv')
df.info()
```

```python
df['horsepower'].fillna(df['horsepower'].mean(), inplace=True)
```

```python
fig=plt.figure(figsize=[6,6])
ax=fig.add_axes([0,0,1,1])
ax.set_title('Distribution of mileage')
ax.boxplot(df['mpg'])
fig.show()
```

```python
df['city_mileage']=df['mpg']*0.75
df.head()
```

```python
mpg_list=[df['mpg'],df['city_mileage']]
fig=plt.figure(figsize=[6,5])
ax=fig.add_axes([0,0,1,1])
ax.set_title('Distribution of mileage')
ax.set_ylabel('Mileage per gallon')
ax.boxplot(mpg_list)
fig.show()
```

```python
grouped_df = df.groupby(['model_year']).count()[['name']]
grouped_df.reset_index(inplace = True)

fig=plt.figure(figsize=[6,5])
ax=fig.add_axes([0,0,1,1])
ax.set_title("Number of cars in each year",fontsize=16)
ax.set_ylabel("Number of cars",fontsize=12)
ax.set_xlabel("Year",fontsize=12)
ax.bar(grouped_df['model_year'], grouped_df['name'])
fig.show()
```

```python
fig=plt.figure(figsize=[8,6])
ax=fig.add_axes([0,0,1,1])
ax.hist(df['horsepower'], bins=20)
ax.set_title('Distribution of horsepower')
ax.set_xlabel('Horsepower',fontsize=12)
ax.set_ylabel('Frequency',fontsize=12)
fig.show()
```

```python
pie_df=pd.DataFrame()
pie_df['Count']=df['origin'].value_counts()
pie_df.reset_index(inplace=True)
pie_df.rename(columns={'index':'Country'}, inplace=True)
fig=plt.figure(figsize=[8,6])
ax=fig.add_axes([0,0,1,1])
ax.set_title('Origin of the cars')
ax.pie(pie_df['Count'], labels=pie_df['Country'])
fig.show()
```
